## Findings — target drift over time (investigate tomorrow)

**The target (`actual_delivery_days`) is non-stationary — it shifts strongly over time.**

- Time-based 80/20 split: **train mean = 12.9 days, test mean = 8.2 days** (a 4.7-day gap; train is the older 80%, test is the most recent 20%).
- Three eras in the monthly average (see the "Avg delivery days by month" chart):
  - **Late 2016:** ~50 days then crashing — Olist launch; very few, erratic orders.
  - **2017:** steady ~12–14 days.
  - **2018:** declining to ~7–8 days — deliveries got faster over time.
- This drift is the cause of the model R² = −2.5 on the held-out test set: it trains on the slow era and is graded on the fast tail, so it over-predicts and scores worse than guessing the mean.
- **Likely right-censoring at the tail:** the dataset ends ~Sept 2018. Orders near the end that took a long time had not been delivered at export, so they have no delivery date and were removed by `dropna`. The latest period is therefore biased toward fast deliveries — part of the 8.2 test mean is artificial, not a real improvement.
